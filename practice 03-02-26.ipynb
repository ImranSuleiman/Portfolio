```python
# Creating new columns 
def age_group(age):
    if age < 25:
        return "18 - 24"
    elif age < 35:
        return "25 -34"
    elif age < 50:
        return "35 -  50"
    else :
        return "50+"
df["Age_Group"] = df["Age"].apply(age_group)
df
```

```python
# Payment method popularity
payment_method = df["Payment Method"].value_counts().sort_values(ascending = False)
payment_method 
```

```python
# Which gender spends the most
df.groupby("Gender")["Purchase Amount (USD)"] .agg(["mean","sum","count"])

# While Men contribute the highest total revenue ($157,890) due to a much larger customer base (2,652 vs 1,248), 
# Women exhibit a slightly higher average transaction value, outspending men by $0.71 per purchase.
```

```python
# Do discounts really increase revenue?
df.groupby("Discount Applied")["Purchase Amount (USD)"].agg(["mean","count"])
# While the discount group has a lower count (suggesting it isn't yet "cannibalizing" all sales), /
# /the lower mean purchase amount indicates that the price cuts are not being offset by customers buying more expensive items or larger quantities. 
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
category_revenue = df.groupby("Category", as_index = False)["Purchase Amount (USD)"].sum().sort_values(by = "Purchase Amount (USD)", ascending=False) # OR .index() - at the end of the code

# Bar chart
sns.barplot( data=category_revenue,
    x="Category",
    y="Purchase Amount (USD)", 
    hue="Category", 
    palette="viridis",  
    legend=False
)
plt.title("Total Revenue by Category")
plt.ylabel("Revenue (USD)")
plt.xlabel("Category")
plt.show()


# other examples of plalettes include 
# palette="deep","pastel","muted","coolwarm","Blues","magma", "rocket", "viridis", "Set2"
```

```python
# Payment method distribution

Payment_method = df["Payment Method"].value_counts()

plt.pie(
    Payment_method.values,
    labels=Payment_method.index,
    autopct="%1.1f%%",
    colors=sns.color_palette("muted")
)
plt.title("Payment Method Popularity")
plt.show()
Payment_method
```

```python
# Revenue by Season
weather_revenue = df.groupby("Season", as_index=False)["Purchase Amount (USD)"].sum().sort_values(by = "Purchase Amount (USD)", ascending =False)

sns.barplot(
    data = weather_revenue,
    x = "Season",
    y = "Purchase Amount (USD)",
    hue = "Season",
    palette = "muted"
)
plt.title("Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue")
plt.show()

```

```python
# Revenue by Gender
gender_revenue = df.groupby("Gender", as_index=False)["Purchase Amount (USD)"].mean().sort_values(by = "Purchase Amount (USD)", ascending =False)

sns.barplot(
    data = gender_revenue,
    x = "Gender",
    y = "Purchase Amount (USD)",
    hue = "Gender",
    palette = "muted"
)
plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Revenue")
plt.show()
```

```python

#===================================
loyal_customer=df.groupby("Loyal Customer")["Purchase Amount (USD)"].mean()
loyal_customer = loyal_customer.reset_index()
sns.barplot(
    data = loyal_customer,
    x="Loyal Customer",
    y="Purchase Amount (USD)",
    hue="Loyal Customer", legend=False,
    palette="Set2"
)
plt.title("Average Spending by Loyalty Status")
plt.ylabel("Average USD")
plt.xlabel("Customer status")
plt.show()
```

## Final Business Insights

1. Discounts, subscriptions, and loyalty do not strongly affect spending.
2. Category is the strongest revenue driver.
   - Clothing dominates total revenue.
   - Footwear has highest average purchase.
   - Outerwear underperforms.
3. Fall is the highest revenue season.
4. Marketing Recommendations:
   - Focus on high-volume categories.
   - Upsell Footwear.
   - Promote Summer to increase off-season sales.
